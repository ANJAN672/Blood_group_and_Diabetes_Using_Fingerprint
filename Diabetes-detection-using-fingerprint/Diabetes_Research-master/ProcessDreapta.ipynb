```python

def  plot_im(im):
    import matplotlib.pyplot as plt
    %matplotlib inline
    import numpy as np
    fig=plt.figure(figsize=(8,6))
    #fig.clf()
    ax=fig.gca()
    ax.imshow(np.array(im).astype(int))
    return 'Done'
```
